```python
import scipy.io
from importlib import resources

# 1. Establish the clean, self-contained absolute file path to the phantom asset
tg119_path = resources.files("pyRadPlan.data.phantoms").joinpath("TG119.mat")

print("--- EXECUTING TRUE ISOLATED DATA CHECK ---")

# 2. Load the raw file
raw_mat_data = scipy.io.loadmat(str(tg119_path), squeeze_me=True, struct_as_record=False)

print("\n--- Core Level Keys in Mat File ---")
print(list(raw_mat_data.keys()))

# 3. Safe attribute extraction check
if 'ct' in raw_mat_data:
    print("\n--- 'ct' key is present! ---")
    ct_struct = raw_mat_data['ct']
    fields = [attr for attr in dir(ct_struct) if not attr.startswith('__')]
    print("Available internal variables:", fields)
else:
    print("\n--- Warning: 'ct' key was NOT found in this .mat dictionary! ---")
```

```python
ct_struct = raw_mat_data['ct']

print("--- Data Array Property Checklist ---")
print(f"cubeHU Data Type: {getattr(ct_struct, 'cubeHU').dtype}")
print(f"cubeHU Matrix Shape: {getattr(ct_struct, 'cubeHU').shape}")

print("\n--- Spatial Tracking Profile ---")
# Let's print out what the resolution property looks like (is it an object or array?)
res = getattr(ct_struct, 'resolution', 'Missing')
if hasattr(res, '__dict__') or isinstance(res, dict):
    print("Resolution attributes:", dir(res) if hasattr(res, 'dir') else res)
else:
    print(f"Resolution value: {res}")

# Check the physical coordinate boundaries
print(f"X coordinate vector sample (first 3 elements): {getattr(ct_struct, 'x')[:3]}")
print(f"Y coordinate vector sample (first 3 elements): {getattr(ct_struct, 'y')[:3]}")
print(f"Z coordinate vector sample (first 3 elements): {getattr(ct_struct, 'z')[:3]}")
```
